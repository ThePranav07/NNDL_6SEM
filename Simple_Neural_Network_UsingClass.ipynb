```python
class FeedForwardNeuralNetwork:
  def __init__(self, input_size, hidden_size, output_size):
    self.w1 = np.random.rand(input_size, hidden_size) * 0.1
    self.b1 = np.zeros(hidden_size)
    self.w2 = np.random.rand(hidden_size, output_size) * 0.1
    self.b2 = np.zeros(output_size)

  def sigmoid(self, x):
    return 1 / (1 + np.exp(-x))

  def forward(self, x):
    self.z1 = np.dot(x, self.w1) + self.b1
    self.a1 = self.sigmoid(self.z1)

    self.z2 = np.dot(self.a1, self.w2) + self.b2
    self.a2 = self.sigmoid(self.z2)
    return self.a2

  def compute_loss(self, y_true, y_pred):
    return (y_true - y_pred)**2

  def backward(self, x, y_true, y_pred, lr):
    dloss_da2 = -2 * (y_true - y_pred)
    da2_dz2 = y_pred * (1 - y_pred)
    dz2_dw2 = self.a1

    grad_w2 = np.dot(dz2_dw2.T, (dloss_da2 * da2_dz2)).reshape(self.w2.shape)
    grad_b2 = (dloss_da2 * da2_dz2).flatten()

    self.w2 -= lr * grad_w2
    self.b2 -= lr * grad_b2

    dloss_da1 = np.dot(dloss_da2 * da2_dz2, self.w2.T)
    da1_dz1 = self.a1 * (1 - self.a1)

    grad_w1 = np.dot(x.T, dloss_da1 * da1_dz1)
    grad_b1 = (dloss_da1 * da1_dz1).flatten()

    self.w1 -= lr * grad_w1
    self.b1 -= lr * grad_b1

  def train(self, X, y, epochs, lr):
    for epoch in range(epochs):
      y_pred = self.forward(X)
      loss = self.compute_loss(y, y_pred)
      self.backward(X, y, y_pred, 1000)
      if (epoch+1) % 100 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.mean()}")
```
