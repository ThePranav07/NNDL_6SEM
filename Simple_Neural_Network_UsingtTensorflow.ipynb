```python
import tensorflow as tf
import numpy as np

model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=2, activation='sigmoid', input_shape=(3,)),
    tf.keras.layers.Dense(units=1, activation='sigmoid')
])

model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.1), loss='mse')

X_train = np.array([[5, 10, 15]], dtype=np.float32)
y_train = np.array([[1.0]], dtype=np.float32)

print("Initial prediction:", model.predict(X_train))

history = model.fit(X_train, y_train, epochs=1000, verbose=0)

print("\nFinal prediction after training:", model.predict(X_train))
print("Final loss:", history.history['loss'][-1])
```
